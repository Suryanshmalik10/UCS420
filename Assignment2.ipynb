Question 1 – Working with Lists

```python
# Q1(i) – Create List
roll=12345
L=[int(x)*10 for x in str(roll)]
print(L)
```

```python
# Q1(ii) – Add Elements
L.append(44)
print(L)
L.insert(0,30)
print(L)
```

```python
# Q1(iii) – Remove Elements
L.remove(44)
print(L)
L.pop()
print(L)
```

```python
# Q1(iv) – Sort Elements
L.sort()
print(L)
L.sort(reverse=True)
print(L)
```

```python
# Q1(v) – Slicing
print(L[:3])
print(L[-3:])
```

```python
# Q1(vi) – List Comprehension
avg=sum(L)/len(L)
new=[x for x in L if x>avg]
print(new)
```

Question 2 – Working with Tuples

```python
# Q2(i) – Highest and Lowest Score
scores=tuple(L[:8])
print(max(scores))
print(scores.index(max(scores)))
print(min(scores))
print(scores.count(min(scores)))
```

```python
# Q2(ii) – Reverse Tuple
result=list(scores[::-1])
print(result)
```

```python
# Q2(iii) – Search Score
x=int(input("Enter score: "))
if x in scores:
    print(scores.index(x))
else:
    print("Not present")
```

```python
# Q2(iv) – Change Tuple Element
scores[0]=100
```

```python
# Q2(v) – Unpacking Tuple
first,second,*remaining=scores
print(first)
print(second)
print(remaining)
```

Question 3 – Random Number Analysis

```python
# Q3 – Random Numbers
import random
random.seed(12345)
L=[random.randint(100,900) for i in range(100)]
print(L)
```

```python
# Q3(i) – Generate 100 Random Numbers
import random
random.seed(12345)
L=[random.randint(100,900) for i in range(100)]
print(L)
```

```python
# Q3(ii) – Odd Numbers
odd=[x for x in L if x%2!=0]
print(len(odd))
print(odd)
```

```python
# Q3(iii) – Even Numbers
even=[x for x in L if x%2==0]
print(len(even))
print(even)
```

```python
# Q3(iv) – Prime Numbers
prime=[x for x in L if x>1 and all(x%i!=0 for i in range(2,int(x**0.5)+1))]
print(len(prime))
print(prime)
```

```python
# Q3(v) – Most Frequent Number
num=max(set(L),key=L.count)
print(num)
print(L.count(num))
```

Question 5 – Working with Dictionaries

```python
# Q5(i) – Rename Key
my_dict={"name":"Suryansh","roll_no":"12345678","branch":"CSE","age":20,"city":"Patiala"}
my_dict["location"]=my_dict.pop("city")
print(my_dict)
```

```python
# Q5(ii) – Add CGPA
my_dict["cgpa"]=9.48
print(my_dict)
```

```python
# Q5(iii) – Update Age
my_dict["age"]=my_dict["age"]+1
print(my_dict)
```

```python
# Q5(iv) – Delete Branch
dict1=my_dict.copy()
dict2=my_dict.copy()
dict1.pop("branch")
del dict2["branch"]
print(dict1)
print(dict2)
```

```python
# Q5(v) – Iterate Over Dictionary
for key,value in my_dict.items():
    print(key+" →",value)
```

```python
# Q5(vi) – Check Email
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email not found")
```

```python
# Q5(vii) – Merge Dictionaries
friend_dict={"name":"Rahul","roll_no":"87654321","branch":"ECE","age":21,"location":"Delhi"}
merged={**my_dict,**friend_dict}
print(merged)
```

```python
# Q5(viii) – Dictionary Comprehension
new_dict={key:value for key,value in my_dict.items() if isinstance(value,str)}
print(new_dict)
```
