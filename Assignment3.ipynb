```python
import pandas as pd

# Q6. Create employees.csv dataset
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df = pd.DataFrame(data)
df.to_csv("employees.csv", index=False)
print("Original Dataset:")
print(df)
```

```python
# a) Shape of DataFrame
print("\na) Shape:")
print(df.shape)
```

```python
# b) Summary: data types and non-null counts
print("\nb) DataFrame Information:")
df.info()
```

```python
# c) Descriptive statistics
print("\nc) Descriptive Statistics:")
print(df.describe())
```

```python
# d) First 5 rows and last 3 rows
print("\nd) First 5 rows:")
print(df.head())

print("\nLast 3 rows:")
print(df.tail(3))
```

```python
# e) Statistics

# i. Average salary
print("\ne-i) Average Salary:", df["Salary"].mean())

# ii. Total bonus
print("e-ii) Total Bonus:", df["Bonus"].sum())

# iii. Youngest employee
print("e-iii) Youngest Age:", df["Age"].min())

# iv. Highest rating
print("e-iv) Highest Rating:", df["Rating"].max())
```

```python
# f) Sort by Salary in descending order
print("\nf) Salary descending:")
print(df.sort_values("Salary", ascending=False))
```

```python

# g) Add Performance Category column
def category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance"] = df["Rating"].apply(category)

print("\ng) Performance Category:")
print(df)
```

```python
# h) Identify missing values
print("\nh) Missing Values:")
print(df.isnull().sum())
```

```python
# i) Rename Employee_ID to ID
df.rename(columns={"Employee_ID": "ID"}, inplace=True)

print("\ni) Renamed Column:")
print(df)
```

```python
# j) Find employees with more than 5 years experience
print("\nj-i) More than 5 years experience:")
print(df[df["Years_of_Experience"] > 5])
```
