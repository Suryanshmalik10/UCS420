5 If-Else - Conditional Checking

Learning: if-else Condition

```python
# Assignment 5.1: WAP to find max among three numbers

a = int(input("Enter First Number: "))
b = int(input("Enter Second Number: "))
c = int(input("Enter Third Number: "))

m = max(a, b, c)

print("Maximum Number =", m)
```

```python
# Assignment 5.2: WAP to add all numbers divisible by 7 and 9 from 1 to n

n = int(input("Enter a number: "))
sum = 0

for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        sum = sum + i

print("Sum =", sum)
```

```python
# Assignment 5.3: WAP to add all prime numbers from 1 to n

n = int(input("Enter a number: "))
sum = 0

for i in range(2, n + 1):
    prime = True

    for j in range(2, i):
        if i % j == 0:
            prime = False
            break

    if prime:
        sum = sum + i

print("Sum of Prime Numbers =", sum)
```

6 Functions Learning: How to declare and call function

6.1 Add two numbers

```python
# Assignment 6.1: WAP using function that add all odd numbers from 1 to n

def addOdd(n):
    sum = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            sum = sum + i

    return sum

n = int(input("Enter a number: "))
print("Sum =", addOdd(n))
```

```python
# Assignment 6.2: WAP using function that add all prime numbers from 1 to n

def addPrime(n):
    sum = 0

    for i in range(2, n + 1):
        prime = True

        for j in range(2, i):
            if i % j == 0:
                prime = False
                break

        if prime:
            sum = sum + i

    return sum

n = int(input("Enter a number: "))
print("Sum =", addPrime(n))
```

16 File Handling

Learning: How to open the file, read the file and write in the file

```python
# 16.1 Write Numbers 1 to 10 in a File

fp = open("result.txt", "w")

for i in range(1, 11):
    fp.write(str(i) + "\n")

fp.close()

print("Writing Done")
```

```python
# 16.2 Read a File and Print Content

fp = open("result.txt", "r")

for line in fp:
    print(line.strip())

fp.close()
```

```python
# 16.3 Read from One File and Write to Another in Uppercase

readfp = open("result.txt", "r")
writefp = open("abc.txt", "w")

for line in readfp:
    writefp.write(line.upper())

readfp.close()
writefp.close()

print("Writing Done")
```
